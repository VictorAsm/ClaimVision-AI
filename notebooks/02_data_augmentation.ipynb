# Notebook 02: Estrategias de Aumento de Datos (Data Augmentation)

## Contexto y Objetivo
En visión computacional aplicada a siniestros automotores, las imágenes tomadas en la vía pública presentan gran variabilidad en iluminación, ángulo de captura, oclusiones y resolución.

Este notebook implementa técnicas de aumento de datos sintéticos sobre las muestras de entrenamiento para:
- Mitigar el sobreajuste (*overfitting*) en categorías de daño minoritarias.
- Robustecer la invariancia espacial y de brillo de los clasificadores convolucionales.


```python
import pandas as pd
from sklearn.model_selection import train_test_split
import config
import os
```

```python
processed_images_dir = F"{config.PROCESSED_PATH_ORIGINAL}/csv"
```

```python
df = pd.read_csv(F"{processed_images_dir}/train.csv")
```

```python
test_set, train_set = train_test_split(df, test_size=0.8, random_state=42)
```

```python
test_set.to_csv(F"{processed_images_dir}/val.csv", index=False)
train_set.to_csv(F"{processed_images_dir}/train.csv", index=False)
```
